```python
# If needed, uncomment:
# %pip install pandas numpy matplotlib scikit-learn

import os, zipfile, warnings
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

warnings.filterwarnings("ignore")
pd.set_option("display.max_columns", 100)
print("Libraries imported successfully.")
```

```python
# Google Colab: upload the downloaded Kaggle ZIP.
# Jupyter/local: set DATA_DIR to your extracted dataset folder.
DATA_DIR = None  # Example: r"C:\Users\Admin\Downloads\stock-data"

try:
    from google.colab import files
    print("Upload the downloaded Kaggle ZIP file.")
    uploaded = files.upload()
    zips = [name for name in uploaded if name.lower().endswith(".zip")]
    if zips:
        DATA_DIR = "/content/stock_dataset"
        Path(DATA_DIR).mkdir(parents=True, exist_ok=True)
        with zipfile.ZipFile(zips[0]) as zf:
            zf.extractall(DATA_DIR)
        print("Extracted to", DATA_DIR)
except ImportError:
    print("Jupyter/local environment detected.")

if DATA_DIR is None:
    DATA_DIR = os.getcwd()
DATA_DIR = Path(DATA_DIR)
print("Dataset folder:", DATA_DIR.resolve())

csv_files = list(DATA_DIR.rglob("*.csv"))
print("CSV files found:", len(csv_files))
if not csv_files:
    raise FileNotFoundError("No CSV files found. Upload/extract the Kaggle dataset and set DATA_DIR correctly.")
print("Examples:", [str(p.relative_to(DATA_DIR)) for p in csv_files[:10]])
```

```python
def norm(s):
    return "".join(c.lower() for c in str(s) if c.isalnum())

DATE_ALIASES = {"date", "datetime", "timestamp", "tradingdate", "pricedate"}
CLOSE_ALIASES = {"close", "closingprice", "adjclose", "adjustedclose", "closeprice", "last", "ltp"}
SYMBOL_ALIASES = {"symbol", "ticker", "stock", "stockname", "company", "tradingsymbol"}
OPEN_ALIASES = {"open", "openingprice"}
HIGH_ALIASES = {"high", "highestprice"}
LOW_ALIASES = {"low", "lowestprice"}
VOLUME_ALIASES = {"volume", "tradedvolume", "totaltradedquantity", "qty"}

def find_col(columns, aliases):
    for col in columns:
        if norm(col) in aliases:
            return col
    for col in columns:
        if any(a in norm(col) for a in aliases if len(a) >= 4):
            return col
    return None

frames, skipped = [], []
for fp in csv_files:
    try:
        raw = pd.read_csv(fp, low_memory=False)
        dc, cc = find_col(raw.columns, DATE_ALIASES), find_col(raw.columns, CLOSE_ALIASES)
        if raw.empty or dc is None or cc is None:
            skipped.append((str(fp), "empty or date/close column not recognized"))
            continue
        out = pd.DataFrame({
            "Date": pd.to_datetime(raw[dc], errors="coerce"),
            "Close": pd.to_numeric(raw[cc].astype(str).str.replace(",", "", regex=False), errors="coerce")
        })
        for name, aliases in [("Open", OPEN_ALIASES), ("High", HIGH_ALIASES),
                              ("Low", LOW_ALIASES), ("Volume", VOLUME_ALIASES)]:
            col = find_col(raw.columns, aliases)
            if col is not None:
                out[name] = pd.to_numeric(raw[col].astype(str).str.replace(",", "", regex=False), errors="coerce")
        sc = find_col(raw.columns, SYMBOL_ALIASES)
        out["Symbol"] = raw[sc].astype(str).str.strip() if sc else fp.stem
        out = out.dropna(subset=["Date", "Close"])
        out = out[out["Close"] > 0]
        if not out.empty:
            frames.append(out)
        else:
            skipped.append((str(fp), "no valid date/close rows"))
    except Exception as e:
        skipped.append((str(fp), str(e)[:150]))

if not frames:
    raise ValueError("No usable CSV files found. Inspect CSV columns and edit DATE_ALIASES/CLOSE_ALIASES.")
data = pd.concat(frames, ignore_index=True, sort=False)
data["Symbol"] = data["Symbol"].replace("", np.nan).fillna("Unknown")
data = data.drop_duplicates(["Symbol", "Date"], keep="last")
data = data.sort_values(["Symbol", "Date"]).reset_index(drop=True)
print("Rows:", f"{len(data):,}", "| Stocks:", data.Symbol.nunique())
print("Overall date range:", data.Date.min(), "to", data.Date.max())
print("Skipped files:", len(skipped))
display(data.head())
```

```python
print("Columns:", list(data.columns))
display(data.select_dtypes(include=np.number).describe().T)
display(data.isna().sum().to_frame("Missing values"))

data = data[(data.Date >= "2000-01-01") & (data.Date <= "2020-12-31")].copy()
data = data.sort_values(["Symbol", "Date"]).reset_index(drop=True)
if data.empty:
    raise ValueError("No rows remain for 2000–2020. Check date parsing and dataset date range.")
print("Rows in 2000–2020:", f"{len(data):,}")
print("Stocks in period:", data.Symbol.nunique())
print("Date range:", data.Date.min().date(), "to", data.Date.max().date())
```

```python
data = data.sort_values(["Symbol", "Date"]).copy()
data["Daily_Return"] = data.groupby("Symbol")["Close"].pct_change()
data["Return_Pct"] = data["Daily_Return"] * 100
data["Market_Move"] = np.select(
    [data.Daily_Return > 0, data.Daily_Return < 0],
    ["Up", "Down"], default="Unchanged"
)
display(data[["Date", "Symbol", "Close", "Daily_Return", "Return_Pct", "Market_Move"]].head(10))

valid_returns = data.dropna(subset=["Daily_Return"])
up = int((valid_returns.Daily_Return > 0).sum())
down = int((valid_returns.Daily_Return < 0).sum())
flat = int((valid_returns.Daily_Return == 0).sum())
n = max(len(valid_returns), 1)
print(f"Up observations: {up:,} ({up/n*100:.2f}%)")
print(f"Down observations: {down:,} ({down/n*100:.2f}%)")
print(f"Unchanged observations: {flat:,} ({flat/n*100:.2f}%)")
print(f"Mean daily return: {valid_returns.Daily_Return.mean()*100:.4f}%")
print(f"Daily return standard deviation: {valid_returns.Daily_Return.std()*100:.4f}%")
```

```python
# Descriptive equal-weighted average across stocks, not an official index.
daily_market = data.groupby("Date", as_index=False).agg(
    Mean_Close=("Close", "mean"),
    Mean_Daily_Return=("Daily_Return", "mean"),
    Stocks_Observed=("Symbol", "nunique")
).sort_values("Date")

plt.figure(figsize=(12, 5))
plt.plot(daily_market.Date, daily_market.Mean_Close)
plt.title("Average Closing Price Across Available Stocks (2000–2020)")
plt.xlabel("Date"); plt.ylabel("Mean close (not an index)")
plt.grid(True, alpha=.3); plt.tight_layout(); plt.show()

plt.figure(figsize=(12, 5))
plt.plot(daily_market.Date, daily_market.Mean_Daily_Return * 100)
plt.axhline(0, linewidth=1)
plt.title("Average Daily Return Across Available Stocks")
plt.xlabel("Date"); plt.ylabel("Mean daily return (%)")
plt.grid(True, alpha=.3); plt.tight_layout(); plt.show()

counts = valid_returns.Market_Move.value_counts().reindex(["Up", "Down", "Unchanged"], fill_value=0)
plt.figure(figsize=(8, 5))
plt.bar(counts.index, counts.values)
plt.title("Up vs Down Stock-Day Observations")
plt.xlabel("Daily move"); plt.ylabel("Number of observations")
plt.tight_layout(); plt.show()
display(daily_market.head())
```

```python
data = data.sort_values(["Symbol", "Date"]).copy()
for lag in [1, 2, 3, 5]:
    data[f"Return_Lag_{lag}"] = data.groupby("Symbol")["Daily_Return"].shift(lag)

data["Return_MA_5"] = data.groupby("Symbol")["Daily_Return"].transform(lambda s: s.rolling(5, min_periods=5).mean())
data["Return_MA_10"] = data.groupby("Symbol")["Daily_Return"].transform(lambda s: s.rolling(10, min_periods=10).mean())
data["Return_STD_5"] = data.groupby("Symbol")["Daily_Return"].transform(lambda s: s.rolling(5, min_periods=5).std())

if {"Open", "High", "Low"}.issubset(data.columns):
    data["Intraday_Range"] = (data.High - data.Low) / data.Close
    data["Open_Close_Change"] = (data.Close - data.Open) / data.Open.replace(0, np.nan)
if "Volume" in data.columns:
    data["Volume_Change"] = data.groupby("Symbol")["Volume"].pct_change().replace([np.inf, -np.inf], np.nan)

data["Target_Next_Day_Return"] = data.groupby("Symbol")["Daily_Return"].shift(-1)
features = ["Daily_Return", "Return_Lag_1", "Return_Lag_2", "Return_Lag_3",
            "Return_Lag_5", "Return_MA_5", "Return_MA_10", "Return_STD_5"]
features += [c for c in ["Intraday_Range", "Open_Close_Change", "Volume_Change"] if c in data.columns]

model_data = data.replace([np.inf, -np.inf], np.nan).dropna(
    subset=features + ["Target_Next_Day_Return", "Date", "Symbol"]
).copy()
print("Features:", features)
print("Usable model rows:", f"{len(model_data):,}")
display(model_data[["Date", "Symbol"] + features + ["Target_Next_Day_Return"]].head())
```

```python
if model_data.Date.nunique() < 10:
    raise ValueError("Not enough distinct dates for a useful chronological train/test split.")

dates = np.array(sorted(model_data.Date.unique()))
split_i = min(max(int(len(dates) * .8), 1), len(dates) - 1)
split_date = pd.Timestamp(dates[split_i])
train = model_data[model_data.Date < split_date]
test = model_data[model_data.Date >= split_date]
if train.empty or test.empty:
    raise ValueError("Empty train/test partition; check dataset size.")

X_train, y_train = train[features], train["Target_Next_Day_Return"]
X_test, y_test = test[features], test["Target_Next_Day_Return"]

model = RandomForestRegressor(n_estimators=150, max_depth=10, min_samples_leaf=5,
                              random_state=42, n_jobs=-1)
model.fit(X_train, y_train)
pred = model.predict(X_test)

baseline = DummyRegressor(strategy="constant", constant=0.0)
baseline.fit(X_train, y_train)
base_pred = baseline.predict(X_test)

def metrics_row(name, y, p):
    return {"Model": name,
            "MAE": mean_absolute_error(y, p),
            "RMSE": np.sqrt(mean_squared_error(y, p)),
            "R2": r2_score(y, p)}

metrics = pd.DataFrame([
    metrics_row("Random Forest", y_test, pred),
    metrics_row("Zero-return baseline", y_test, base_pred)
])
print("Train period:", train.Date.min().date(), "to", train.Date.max().date())
print("Test period:", test.Date.min().date(), "to", test.Date.max().date())
print("Train rows:", f"{len(train):,}", "| Test rows:", f"{len(test):,}")
display(metrics)
```

```python
comparison = test[["Date", "Symbol", "Close", "Target_Next_Day_Return"]].copy()
comparison["Predicted_Next_Day_Return"] = pred
comparison["Actual_Return_Pct"] = comparison.Target_Next_Day_Return * 100
comparison["Predicted_Return_Pct"] = comparison.Predicted_Next_Day_Return * 100
display(comparison.head(20))

sample = comparison.sort_values("Date").head(250)
plt.figure(figsize=(12, 5))
plt.plot(sample.Date, sample.Actual_Return_Pct, label="Actual", linewidth=1)
plt.plot(sample.Date, sample.Predicted_Return_Pct, label="Predicted", linewidth=1)
plt.axhline(0, linewidth=1)
plt.title("Actual vs Predicted Next-Day Returns (Test Sample)")
plt.xlabel("Date"); plt.ylabel("Return (%)"); plt.legend()
plt.grid(True, alpha=.3); plt.tight_layout(); plt.show()

importance = pd.Series(model.feature_importances_, index=features).sort_values()
importance.plot(kind="barh", figsize=(8, 5), title="Random Forest Feature Importance")
plt.xlabel("Importance"); plt.tight_layout(); plt.show()
display(importance.sort_values(ascending=False).to_frame("Importance"))
```

```python
latest = data.sort_values(["Symbol", "Date"]).groupby("Symbol", as_index=False).tail(1).copy()
latest = latest.replace([np.inf, -np.inf], np.nan).dropna(subset=features)
if latest.empty:
    print("No stock has enough historical data for prediction.")
    next_predictions = pd.DataFrame()
else:
    latest["Predicted_Next_Return_Pct"] = model.predict(latest[features]) * 100
    next_predictions = latest[["Date", "Symbol", "Close", "Predicted_Next_Return_Pct"]].sort_values(
        "Predicted_Next_Return_Pct", ascending=False)
    print("Top 10 estimated next returns:")
    display(next_predictions.head(10))
    print("Bottom 10 estimated next returns:")
    display(next_predictions.tail(10))
```

```python
out_dir = Path("stock_market_outputs")
out_dir.mkdir(exist_ok=True)
data.to_csv(out_dir / "cleaned_stock_data_with_returns.csv", index=False)
daily_market.to_csv(out_dir / "daily_market_summary.csv", index=False)
comparison.to_csv(out_dir / "test_predictions.csv", index=False)
metrics.to_csv(out_dir / "model_metrics.csv", index=False)
if not next_predictions.empty:
    next_predictions.to_csv(out_dir / "next_return_predictions.csv", index=False)
print("Saved output files in:", out_dir.resolve())
print([p.name for p in out_dir.iterdir()])
```
